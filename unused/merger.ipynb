```python
from PIL import Image
import matplotlib.pyplot as plt
import os

# Your 6 plot filenames
images = [
    "bone.png",
    "brain.png",
    "chest.png",
    "eurosat.png",
    "neu.png",
    "output.png"
]

# Folder containing the images
folder = r"C:/Users/sunda/Downloads"

# Create 2 × 3 layout
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()

labels = ["(a)", "(b)", "(c)", "(d)", "(e)", "(f)"]

# Load and display images
for i, image_name in enumerate(images):
    image_path = os.path.join(folder, image_name)
    img = Image.open(image_path)

    axes[i].imshow(img)
    axes[i].set_title(labels[i], fontsize=14)
    axes[i].axis("off")

plt.tight_layout()

# Save merged figure
plt.savefig("merged_6_plots.png", dpi=300, bbox_inches="tight")

plt.show()
```
